## O Algoritmo KNN (K Nearest Neighbor)


O algoritmo K-Nearest Neighbor (KNN) é uma técnica de aprendizado de máquina utilizada para tarefas de classificação e regressão. Ele funciona identificando os K pontos de dados mais próximos de uma determinada entrada e realizando previsões com base na classe majoritária ou no valor médio desses vizinhos.
 O algoritmo:
* Classifica os dados baseados na similaridade com pontos de dados próximos
* Usa métricas como a distância Euclidiana para achar os vizinhos mais próximos


No algoritmo KNN, o *k* é simplesmente um número que indica ao algoritmo quantos pontos próximos — ou vizinhos — ele deve considerar ao tomar uma decisão.
Exemplo: imagine que você está tentando identificar uma fruta com base em seu formato e tamanho. Você a compara com frutas que já conhece. 

Se *k* = 3, o algoritmo analisa as 3 frutas mais próximas da nova fruta. 
Se 2 dessas 3 frutas forem maçãs e 1 for uma banana, o algoritmo classifica a nova fruta como maçã, pois a maioria de seus vizinhos são maçãs.
  

```python
import numpy as np
from collections import Counter
```

O método `Counter` é utilizado para contar as ocorrências de elementos em uma lista ou iterável. No KNN, após encontrar os rótulos dos *k* vizinhos mais próximos, o `Counter` ajuda a contabilizar quantas vezes cada rótulo aparece.

```python
# A métrica usada é a distância Euclidiana
def euclidean_distance(point1, point2):
    return np.sqrt(np.sum((np.array(point1) - np.array(point2))**2))
```

Implementação da Função de Predição KNN

* `distances.append` armazena a distância de cada ponto de treinamento em relação ao ponto de teste, juntamente com seu rótulo.
* `distances.sort` é utilizado para ordenar a lista, de modo que os pontos mais próximos apareçam primeiro.
* `k_nearest_labels` seleciona os rótulos dos *k* pontos mais próximos.
* Utiliza o `Counter` para identificar qual rótulo ocorre com maior frequência entre esses *k* rótulos, definindo-o como a predição.

```python
def knn_predict(training_data, training_labels, test_point, k):
    distances = []
    for i in range(len(training_data)):
        dist = euclidean_distance(test_point, training_data[i])
        distances.append((dist, training_labels[i]))
    distances.sort(key=lambda x: x[0])
    k_nearest_labels = [label for _, label in distances[:k]]
    return Counter(k_nearest_labels).most_common(1)[0][0]
```

* `distances.append((dist, training_labels[i]))` cria uma Tupla guardando a distância de cada ponto e seu *label*, ou classe.
* `key=lambda x: x[0]`: key retorna o primeiro elemento de cada Tupla, que será utilizado como referência para ordenamento do conjunto. A expressão lambda x: x[0] é equivalente a:

```python
def chave(x):
    return x[0]
```

`distances.sort(key=lambda x: x[0])` retorna a lista com ordem crescente de distâncias.  Como exemplo, imagine um conjunto de pontos se objetos cujas classes sejam cores, e a lista `distances`é :

```python
[
 (2.8,"Azul"),
 (1.2,"Azul"),
 (4.1,"Vermelho"),
 (0.9,"Azul")
]
```

Após `distances.sort(key=lambda x: x[0])` teremos:

```python
[
 (0.9,"Azul"),
 (1.2,"Azul"),
 (2.8,"Azul"),
 (4.1,"Vermelho")
]
```

O comando `k_nearest_labels = [label for _, label in distances[:k]]` extrai apenas a classe dos *k* pontos mais próximos , ignorando as distâncias.
`Counter(k_nearest_labels)`conta quantas vezes cada rótulo aparece. Suponha que *k=3*. Então  

```python
Counter({
    "Azul":2,
    "Vermelho":1
})
```

`.most_common(1)`retorna `[("Azul",2)]`, onde `(1)`informa que deseja-se apenas 1 elemento, o mais comum. 

`Counter(k_nearest_labels).most_common(1)[0]` retorna o primeiro par da Tupla, ou seja, `("Azul",2)`, e 

`Counter(k_nearest_labels).most_common(1)[0][0]`retorna o primeiro elemento do par, isto é, `Azul`. 

Dados de Treino, *Labels* (Rótulos ou Classes)e Ponto Teste:

```python
training_data = [[1, 2], [2, 3], [3, 4], [6, 7], [7, 8]]
training_labels = ['A', 'A', 'A', 'B', 'B']
test_point = [4, 5]
k = 3
```

Predição:

```python
prediction = knn_predict(training_data, training_labels, test_point, k)
print(prediction)
```

O algoritmo calcula as distâncias do ponto de teste [4, 5] em relação a todos os pontos de treinamento, seleciona os 3 pontos mais próximos (considerando k = 3) e determina seus rótulos. Como a maioria dos pontos mais próximos possui o rótulo 'A', o ponto de teste é classificado como 'A'.

Desvantagens do KNN:

* Lentidão com grandes volumes de dados: requer a comparação de cada ponto durante a previsão. 
* Dificuldade com muitas características: a precisão diminui quando os dados possuem um número excessivo de características. 
* Risco de *overfitting*: o KNN pode sofrer *overfitting* quando o valor de *k* é muito baixo ou quando os dados contêm ruído.
