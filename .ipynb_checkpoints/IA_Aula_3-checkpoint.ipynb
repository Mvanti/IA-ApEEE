## Exercícios com Python e Numpy


```python
import numpy as np
```

Em Python, um arranjo NumPy é uma coleção modificável de elementos numéricos homogêneos eficiente numericamente.

```python
l1=[1,2,3]
l2=[4,5,6]
a1=np.array(l1)
a2=np.array(l2)
print(a1.dot(a2))
```

* 1. Crie um arranjo 1D com valores entre 0 10.

```python
arr=np.arange(10)
print(arr)
```

* 2. Converta o arranho 1D criado em um arranjo 2D (matriz) com duas linhas

```python
arr=arr.reshape(2,-1)
print(arr)
```

* 3. Multiplique uma matriz 5x3por uma matriz 3x2, criadas com valores randômicos.

```python
M1=np.random.random((5,3))
M2=np.random.random((3,2))
print('M1=',M1)
print('M2=',M2)
Mres=np.dot(M1,M2)
print('produto=',Mres)
```

* 4. Extraia todos números ímpares de um arranjo 1-10

```python
arr=np.arange(1,11)
impar=arr[arr%2!=0]
print(impar)
```

* 5. Crie uma matriz 3x3 com elementos aleatórios e calcule a soma dos elementos da diagonal.

```python
M=np.random.random((3,3))
soma_diag=np.trace(M)
print(soma_diag)
```

* 6. Tranforme o arranjo 1D com elementos 2,1,4.5,7.8,9,10,2,2.1,3.1,0 em uma matriz com 5x2.

```python
arr=np.array([2,1,4.5,7.8,9,10,2,2.1,3.1,0])
print(len(arr))
print(arr.shape)
M=arr.reshape(-1,2) # teste com (5,-1)
print(M)
```

* 7. Empilhe dois arranjos 1D de 3 elementos verticalmente, cormando uma matriz 2x3.

```python
a1=np.array([1,2,3])
a2=np.array([4,5,6])
M=np.vstack((a1,a2))
print(M)
```

* 9. Em Python uma Lista é uma coleção de elementos modificável. crie uma lista com elementos de tipos distintos, e modifique um deles. 

```python
lst=['casa', 2, np.pi, False] # Lista com uma string, um inteiro, um real irracional e uma variável boleana. 
print(type(lst))
print(lst)
lst[2]=np.exp(1)
print(lst)
lst.append(1)
lst.append(True)
print(lst)
```

* 11. Crie uma matriz 3x3 com elementos 1-9

```python
M=np.arange(1,10).reshape(3,3) # mesmo que (3,-1) OU (-1,3)
print(M)
```

* 12. Crie uma matrix 3x3 com 5 na diagonal principal, transponha e inverta a matriz criada.

```python
M=np.eye(3)*5
print(M)
print(M.T)
print(np.linalg.inv(M))
```
