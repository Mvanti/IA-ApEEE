Por outro lado, O NumPy (Numerical Python) é baseado no objeto ndarray que é muito mais eficiente para cálculos numéricos. Ao contrário das listas padrão do Python, os arrays do NumPy suportam operações rápidas sobre os elementos em paralelo, sem a necessidade de loops explícitos. Isso torna os cálculos significativamente mais rápidos e eficientes, especialmente para grandes conjuntos de dados.

```python
import time
import numpy as np

# criação de listas em grandes conjuntos de dados
size = 10000000
x_G = list(range(size))
y_G = list(range(size))

start = time.time()
z_G = [x_G[i] + y_G[i] for i in range(size)]
end = time.time()
print("Python list computation time:", end - start)

# implementação NumPy 
x_G_np = np.array(x_G)
y_G_np = np.array(y_G)

start = time.time()
z_G_np = x_G_np + y_G_np
end = time.time()
print("NumPy array computation time:", end - start)
```

Indexação e Slicing em NumPy através de exemplos:

```python
import numpy as np

arr1 = np.arange(10, 1, -2)
print("Array:", arr1)

arr2 = arr1[[3, 1, 2]]
print("Elementos índices específicos:", arr2)
```

Slicing: x[start:end:step], onde:

* start: índice do elemento inicial incluído
* stop: Indice final não incluído
* step: intervalo entre elementos

```python
## import numpy as np  
a = np.arange(20)
print("a[:]=",a[:])
print("a[15] =", a[15]) 
print("a[:15]=",a[:15]) # 0-14
print("a[15:] =", a[15:])# 15-19
print("a[-1:20] =", a[-1:20])# último elemento
print("a[-5:18]=", a[-5:18]) # quinto, a partir do último, até o décimo sétimo. 
print("a[::-1]=",a[::-1]) # start e stop não são fornecidos e os valores linmites são assumidos . Como step=-1, o arrajo é revertido. 
print("a[19:5:-1]=",a[19:5:-1]) 
```

Seleção das linhas em uma matriz cuja soma dos elementos é um múltiplo de 10:

```python
b = np.array([[5, 5], [4, 5], [16, 4]])
res = b.sum(axis=1) # soma realizada ao longo do segundo eixo:
print(res)
print(b[res % 10 == 0,:])
```

Agora selecione as colunas cuja soma é múltiplo de 5:

```python
res=b.sum(axis=0) # soma realizada ao longo do primeiro eixo:
print(res)
print(b[:,res % 5 == 0])
```

Exercício: Análise Matricial de Medições- Escreva uma rotina em Python, utilizando NumPY, para analisar a seguinte matriz: 

```python
X = np.array([
    [12, 15, 18, 20, 22],
    [10, 14, 16, 18, 20],
    [20, 18, 15, 12, 10],
    [ 8, 10, 12, 14, 16]
])
```

A rotina deve:
* Calcular as dimensões da matriz
* Calcular o vetor contendo a soma das medições em cada instante, por equipamento
* Calcular a média das medições de cada equipamento
* Selecionar os equipamentos cuja soma das medições seja superior a 75
* Selecionar as colunas cuja soma seja múltipla de 2
* Multiplicar a matriz pelo vetor de ponderação p = np.array([0.8, 1.0, 1.2, 1.1, 0.9])
* Normalização por colunas: Cada coluna deve ser normalizada utilizando:
  $$X_{ij}=\frac{X-\mu}{\sigma}$$

```python
X.shape
```

A matriz possui 4 equipamentos com 5 instantes de medição.

```python
# soma de todas as medidas por equipamento
soma_med=X.sum(axis=1)
print(soma_med)
```

```python
# Média por equipamento:
med_equip= X.mean(axis=1)
print(med_equip)
```

```python
res=X[soma_med>75,:]
print(res)
```

```python
soma_col=X.sum(axis=0)
X[:,soma_col %2 ==0]
```

```python
p = np.array([0.8, 1.0, 1.2, 1.1, 0.9])
y=X@p # equivalente a X.dot(p)
y
```

```python
media = X.mean(axis=0)
desvio = X.std(axis=0)

X_normalizado = (X - media) / desvio
print("Média=",media)
print("Desvio Padrão=",desvio)
print("Matrix X Normalizada=", X_normalizado)
```
